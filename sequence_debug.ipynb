```python
import torch

from src.transforms.multi_layer_parity_conv1d.tanh import MultiLayerTanhParityConv1DTransform
from src.transforms.normalization.actnorm import ActNorm
from src.transforms.parity_conv1d import ParityConv1DTransform
```

```python
data = torch.randn(10, 3, 100)

K = 7
L = 32
```

```python
# check invertibility of MultiLayerTanhParityConv1DTransform
transform = MultiLayerTanhParityConv1DTransform(3, 100, K, L, order=2, parity=0)
transformed_data, _ = transform(data)
reconstructed_data, _ = transform.inverse(transformed_data)
print("error,", torch.norm(data - reconstructed_data).item())
```

```python
# check invertibiilty of ActNorm
actnorm = ActNorm(3)
transformed_data, _ = actnorm(data)
reconstructed_data, _ = actnorm.inverse(transformed_data)
print("error,", torch.norm(data - reconstructed_data).item())
```

```python
# check invertibility of ParityConv1DTransform
parity_conv1d = ParityConv1DTransform(3, 100, K, parity=0)
transformed_data, _ = parity_conv1d(data)
reconstructed_data, _ = parity_conv1d.inverse(transformed_data)
print("error,", torch.norm(data - reconstructed_data).item())
```
